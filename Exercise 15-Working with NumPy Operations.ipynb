## Ex1 -Business Analytics: Sales Performance Calculator.

#### Task 1: Create the Dataset

```python
# the code below is to import NumPy and create the dataset needed for this exercise.

import numpy as np

# the code below is the provided dataset of daily sales record of a business firm.

daily_sales = np.array([125000, 150000, 175000, 140000, 190000, 210000, 160000])

# the code below is to display the array (daily_sales), the datatype of the array and
# the shape of the array.

print ("Daily Sales Record:", daily_sales)
print ("Data Type:", type(daily_sales))
print ("Array Shape: 1-Dimensional Array:", daily_sales)
```

#### Task 2: Calculate Total Sales.

```python
# the code below is calculate the total sales record from the daily_sales dataset.

total_sales = np.sum(daily_sales)

print ("Total Sales:", total_sales)
```

#### Task 3: Calculate Average Daily Sales.

```python
# the code below is to calculate the average of the daily sales.
# we will do this using the np.mean() function

average_sales = np.mean(daily_sales)

print ("Average Sales:",round(average_sales,2))
print ('-' *70)
print (f"""The average sales tells the firm if they are behind schedule or 
above schedule in terms of the expected/fixed sales target per day/week""")
```

#### Task 4: Apply A Sales Adjustment

```python
# the code below is to represent the adjusted sales record with a 10% increment for estimation purposes.
# we will solve this task by first getting the 10% of each sales record,
# then we add it to the daily_sales record.

increment = daily_sales * (10/100)
adjusted_sales = daily_sales + increment

print ("Total Adjusted Sales Record:", adjusted_sales)
```

#### Task 5: Compare Original and Adjusted Sales record.

```python
# the code below is to compare and display the difference between the original and adjusted sales record.

sales_difference = adjusted_sales - daily_sales

print ("Estimated Sales Difference:", sales_difference)
```

## Ex2: Education Analysis - Student Performance Analysis

#### Task 1: Create and Inspect Data

```python
# the code below is to create the dataset needed for this exercise, display the dataset, the type and its shape.

scores = np.array([62, 75, 81, 69, 88, 94, 73, 85, 77, 91])
print ("Scores:", scores)
print('-' * 50)

# to display the datatype
print ("Scores Datatype:", type(scores))
print('-' * 50)

# to display the shape
print ("Shape: 1-Dimensional:", scores)
```

#### Task 2: Calculate the mean score

```python
# the code below is to calculate and display the mean score of the student scores.

mean_score = np.mean(scores)
print ("Mean score:", mean_score)
```

#### Task 3: Calculate each Student's Deviation

```python
# the code below is to calculate and display the deviation of each student's score.
# to get the deviation of each student score, we minus the mean from the student score.

deviation = scores - mean_score
print ("Student Deviation:", deviation)
```

#### Task 4: Identify Positive and Negative Deviation

##### question: 
##### 1: What does a positive deviation mean?
##### 2: What does a negative deviation mean?
##### 3: What does a value close to zero mean?

##### Answer:
##### 1: A positive deviation means the student scored above the pass threshold.
##### 2: A negative deviation means the student scored below the pass threshold.
##### 3: A Value close to Zero means the student was positively or negatively close to the pass threshold.

#### Task 5: Square the Deviations

```python
# the code below is to square the deviations and display the results

squared_deviation = deviation ** 2

print ("Squared Deviation:", squared_deviation)
print ('-' * 70)

print (f"""Why squaring the negative score values changes them to positive scores...
Based on the rule of mathematical operation, negative multilied by negative gives you a postive value,
so therefore, multiplying the negative scores by itself (which is the meaning of squaring the value)
converts and returns the result in a positive score value.""")
```

#### Task 6: Calculate the sum of Squared Deviations

```python
# the code below is to calculate the sum of the squared deviations and display the value.

sum_of_squared_deviations = np.sum(squared_deviation)
print ("Sum of Squared Deviations:",sum_of_squared_deviations)
```

#### Task 7: Connect the Steps to Standard Deviation
##### Standard Deviation Steps
Calculate the mean --> Calculate deviations --> Square the deviations --> Sum the squared deviations --> calculate the standard deviation


```python
# the code below is to connect the steps to standard deviation, perform the operation and display the result.

# the scores
scores

# the mean of the scores
mean_score

# the deviation
deviation

# the squared deviations
squared_deviation

# the sum of the squred deviations
sum_of_squared_deviations

# to get number of student scores recorded 
N = len(scores)

# the standard deviation
standard_deviation = np.sqrt( 1/(N-1) * sum_of_squared_deviations)

print ("Standard Deviation of the Scores:", round(standard_deviation,2))
```

The Standard Deviation is useful when analysing student scores because
it helps the analyst to know more details regarding the score that the mean alone cannot tell.
The mean practically tells where the student score is centered, that is, the average...
but then, the standard deviation helps us to understand if the students score is relatively consistent
to the mean score or whether the situations is actually that some student are performing very well,
while some others are actually struggling.

##### Why Looking beyond the Average Score telld more about the student score.

As I have explained above, looking below the mean score of the student will help the examiners understand
whether the class as a whole is performing well or poorly.
OR
if among the students, some are performing very highly, while the others are performing very poorly and actually need more attention.

## Ex 3: Engineering and Scientific Computing - Trigonometric Series and Convergence.

#### Task 1: Define the Angle.

```python
# the code below is to assign angle 30 to the variable theta and convert it to radians.

theta = 30
theta_radians = np.radians(theta)

print ("Angle in Radians:", round (theta_radians,1))
```

#### Task 2: Create the Sequence of Terms

```python
# the code below is to create a sequence of values

k = np.arange(1, 100, 1)

print ("K-Sequence:", k)
print ('-' * 70)
print ("first value in the sequence:", k[0])
print ('-' * 70)
print ("last value in the sequence:", k[-1])
print ('-' * 70)
print ("Number of Values generated:", len(k))
```

#### Task 3: Calculate the trigonometric Component

```python
# the code below is to calculate and display the trigonometric component of the sequence.

trigonometric_component = t_c = np.sin(theta_radians)
print (t_c)
```

#### Task 4: Build the Series Terms

```python
# the code below is to create and display the series terms.

terms = np.sin(theta_radians)/k

print ("Terms:", terms)
```

#### Task 5: Calculate the Sum

```python
# the code below is to calculate the sum of the generated terms.
total_terms = np.sum(terms)

print ("Total Terms:", total_terms)
```

#### Task 6: Increase the number of Terms

```python
# for when len(k) = 99
k = np.arange(1,100,1)
terms = np.sin(k + theta_radians)/k
total_terms1 = np.sum(terms)
print ("Series Result when K = 99:",total_terms1)
print ()

# for when len(k) = 999
k = np.arange(1,1000,1)
terms = np.sin(k + theta_radians)/k
total_terms2 = np.sum(terms)
print ("Series Result when K = 999:",total_terms2)
print ()

# for when len(k) = 9999
k = np.arange(1,10000,1)
terms = np.sin(k + theta_radians)/k
total_terms3 = np.sum(terms)
print ("Series Result when K = 9,999:",total_terms3)
print ()

```

#### Task 7: Investigate the Convergence

```python
# the code below is to compare the series results for the three values of len(k)

print (f"""
=========================================
|    NUMBER OF TERMS | SERIES RESULT    |
=========================================
|    99              |0.939234003522875 |
=========================================
|    999             |0.9477794883282241|
=========================================
|    9,999           |0.9484451042034616|
=========================================
""")
print()
print ("The series result came closer to becoming 1 as the number of terms increased.")
```

##### 4: What does it mean when the result gets closer to a limiting value?

##### Answer:

##### 3: Increasing the number of terms changes the approximate result because every additional term contributes something to the total.

##### 4: When we say a result gets closer to a limiting value, it means that as you increase the number of terms or iterations, the calculated result approaches a particular number and the changes become smaller and smaller.
